```python
text = input("Enter a string: ")
count = 0

for ch in text:
    if ch.isupper():
        count += 1

print("Number of capital letters:", count)
```

Part one explanation:

* Loop through each character.

* isupper() checks if character is uppercase.

* Increase counter if true.

```python
text = input("Enter a string: ")
vowels = "aeiouAEIOU"
count = 0

for ch in text:
    if ch in vowels:
        count += 1

print("Total vowels:", count)
```

Part two explanation:

* Store vowels in a string.

* Check if character exists using in.

* Count occurrences.

```python
sentence = input("Enter a sentence: ")

words = sentence.split()

for word in words:
    print(word)
```

Part three explanation:

* split() breaks sentence into words.

* Loop prints each word on new line.

```python
string = input("Enter string: ")
substring = input("Enter substring: ")

count = 0

for i in range(len(string) - len(substring) + 1):
    if string[i:i+len(substring)] == substring:
        count += 1

print(count)
```

Part four explanation:

* Traverse string from left to right.

* Compare sliced part with substring.

* Increase count when matched.

```python
text = input("Enter string: ")
text = text.upper()

checked = set()

for ch in text:
    if ch.isalpha() and ch not in checked:
        count = text.count(ch)
        print(count, ch)
        checked.add(ch)
```

Part five explanation:

* Convert string to uppercase (case insensitive).

* Use set() to avoid repeating letters.

* count() counts occurrences.

```python
sentence = input("Enter sentence: ")

words = sentence.split()
unique_words = set(words)

print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
```

Part six explanation:

* Set automatically removes duplicates.

* len() gives count

```python
n = int(input("Enter number of fruits: "))

s1 = set()
s2 = set()

print("Enter fruits for set 1:")
for i in range(n):
    s1.add(input())

print("Enter fruits for set 2:")
for i in range(n):
    s2.add(input())

print("Common fruits:", s1 & s2)
print("Only in s1:", s1 - s2)
print("Total fruits:", len(s1 | s2))
```

Part seven explanation:

* Intersection (&) → Finds fruits common in both sets.

* Difference (-) → Finds fruits present in s1 but not in s2.

* Union (|) → Combines both sets and removes duplicates.

* len() is used to count total unique fruits.

```python
S1 = {"Red", "yellow", "orange", "blue"}
S2 = {"violet", "blue", "purple"}

print("Union:", S1 | S2)
print("Intersection:", S1 & S2)
print("Difference S1-S2:", S1 - S2)
print("Difference S2-S1:", S2 - S1)
print("Symmetric Difference:", S1 ^ S2)
```

Part eight explanation:

* Union (|) → Combines all elements from both sets without duplicates.

* Intersection (&) → Finds common elements in both sets.

* Difference (-) → Finds elements present in one set but not in the other.

* Symmetric Difference (^) → Finds elements present in either set but not in both.
