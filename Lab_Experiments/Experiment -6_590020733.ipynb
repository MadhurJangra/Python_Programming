```python
n = int(input("Enter number of values: "))
values = []

for i in range(n):
    num = int(input("Enter value (0-3): "))
    values.append(num)

for i in range(4):
    print("Number", i, "occurred", values.count(i), "times")
```

Part one explanation:

* Store inputs in list.

* Use count() to count occurrences.

* Loop from 0 to 3 to print frequency.

```python
n = int(input("Enter number of elements: "))
numbers = []

for i in range(n):
    num = float(input("Enter number: "))
    numbers.append(num)

t = tuple(numbers)

average = sum(t) / len(t)

print("Tuple:", t)
print("Average:", average)
```

Part Two explanation:

* Take input in list.

* Convert list to tuple using tuple().

* Use sum() and len() to calculate average.

```python
n = int(input("Enter number of persons: "))
students = {}

for i in range(n):
    name = input("Enter name: ")
    city = input("Enter city: ")
    students[name] = city


print("Names:", students.keys())

print("Cities:", students.values())

for name, city in students.items():
    print(name, "lives in", city)

city_count = {}
for city in students.values():
    city_count[city] = city_count.get(city, 0) + 1

print("Students in each city:", city_count)
```

Part four explanation:

* Dictionary stores data as key: value.

* .keys() → names

* .values() → cities

* .items() → both

* .get() safely updates count.

```python
n = int(input("Enter number of movies: "))
movies = {}

for i in range(n):
    name = input("Movie name: ")
    year = int(input("Release year: "))
    director = input("Director name: ")
    cost = float(input("Production cost: "))
    collection = float(input("Collection made: "))

    movies[name] = {
        "year": year,
        "director": director,
        "cost": cost,
        "collection": collection
    }

for name, details in movies.items():
    print(name, details)

print("Movies released before 2015:")
for name, details in movies.items():
    if details["year"] < 2015:
        print(name)

print("Profitable movies:")
for name, details in movies.items():
    if details["collection"] > details["cost"]:
        print(name)

search_director = input("Enter director name to search: ")
for name, details in movies.items():
    if details["director"] == search_director:
        print(name)
```

Part five explanation:

* Outer dictionary stores movie name.

* Inner dictionary stores movie details.

* Compare collection > cost → profit.

* Check year < 2015 → older movies.

* Search by director name.
