Part one step one explanation:
* "w" → write mode (creates/overwrites file)

```python
with open("name.txt", "r") as file:
    names = file.readlines()
    print("Total names:", len(names))
```

Part one a part explanation:
* "r" → read mode
* readlines() → returns list of lines
* len(names) → counts total names


```python
count = 0
vowels = "AEIOUaeiou"

with open("name.txt", "r") as file:
    for name in file:
        if name.strip()[0] in vowels:
            count += 1

print("Names starting with vowel:", count)
```

Part one b part explanation:
* strip() → removes spaces/newline
* [0] → first character
* Check if first letter is vowel

```python
longest = ""

with open("name.txt", "r") as file:
    for name in file:
        name = name.strip()
        if len(name) > len(longest):
            longest = name

print("Longest name:", longest)
```

Part one c part explanation:
* Compare length of each name
* Store longest found

```python
with open("numbers.txt", "w") as file:
    file.write("10\n150\n200\n50\n75\n120\n")
```

```python
with open("numbers.txt", "r") as file:
    nums = [int(x.strip()) for x in file]

print("Max number:", max(nums))
```

Part two part a explanation:
* Convert each line to integer
* Use max() to get largest

```python
avg = sum(nums) / len(nums)
print("Average:", avg)
```

Part two part b explanation:
* sum(nums) → total
* len(nums) → count
* Average = sum / count

```python
count = 0

for num in nums:
    if num > 100:
        count += 1

print("Numbers > 100:", count)
```

Part two part c explanation:
* Loop through list
* Check condition and increment counter

```python
with open("city.txt", "r") as file:
    for line in file:
        print(line.strip())
```

Part three part a explanation:

```python
with open("city.txt", "r") as file:
    for line in file:
        city, pop, area = line.split()
        if float(pop) > 10:
            print(city)
```

Part three part b explanation:
* split() → separates values
* Convert pop to float
* Check condition

```python
total_area = 0

with open("city.txt", "r") as file:
    for line in file:
        city, pop, area = line.split()
        total_area += float(area)

print("Total area:", total_area)
```

Part three part c explanation:
* Extract area
* Convert to float
* Add to total

```python
n = int(input("Enter number of test cases: "))

for i in range(n):
    try:
        a, b = input().split()
        result = int(a) // int(b)
        print(result)

    except ZeroDivisionError as e:
        print("Error Code:", e)

    except ValueError as e:
        print("Error Code:", e)
```

Part four explanation:
* try → code that may cause error
* except → handles errors

```python
class FileEmptyError(Exception):
    pass

class InvalidDataError(Exception):
    pass

try:
    with open("numbers.txt", "r") as file:
        data = file.read().strip()

        if data == "":
            raise FileEmptyError("File is empty")

        for line in data.split("\n"):
            if not line.isdigit():
                raise InvalidDataError("Invalid data found")

except FileNotFoundError:
    print("File not found")

except FileEmptyError as e:
    print(e)

except InvalidDataError as e:
    print(e)
```
