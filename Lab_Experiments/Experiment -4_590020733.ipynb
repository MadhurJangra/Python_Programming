```python
num = int(input("Enter number: "))
temp = num
sum = 0

while temp > 0:
    digit = temp % 10
    sum += digit ** 3
    temp //= 10

if sum == num:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
num = int(input("Enter number: "))

if num > 1:
    for i in range(2, num):
        if num % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime Number")
else:
    print("Not Prime")
```

Part four explanation:

* Prime → divisible only by 1 and itself

* Check divisibility from 2 to num-1

* If divisible → not prime

```python
num = int(input("Enter number: "))
temp = num
reverse = 0

while temp > 0:
    digit = temp % 10
    reverse = reverse * 10 + digit
    temp //= 10

if reverse == num:
    print("Palindrome")
else:
    print("Not Palindrome")
```

Part five explanation:

* Reverse the number

* Compare reversed with original

* If same → palindrome

```python
num = int(input("Enter number: "))
sum = 0

while num > 0:
    digit = num % 10
    sum += digit
    num //= 10

print("Sum of digits:", sum)
```

Part seven explenation:

* Loop from 1 to 100

* Use or operator

* Count how many numbers satisfy condition

```python
text = input("Enter string: ")

result = ""

for ch in text:
    if ch >= 'a' and ch <= 'z':
        result += chr(ord(ch) - 32)
    else:
        result += ch

print("Uppercase string:", result)
```
