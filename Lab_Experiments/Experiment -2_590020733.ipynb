```python
import math

a = float(input("Enter side 1: "))
b = float(input("Enter side 2: "))
c = float(input("Enter side 3: "))

s = (a + b + c) / 2
area = math.sqrt(s * (s - a) * (s - b) * (s - c))

print("Area of triangle:", area)
```

```python
seconds = int(input("Enter seconds: "))

hours = seconds // 3600
minutes = (seconds % 3600) // 60
remaining_seconds = seconds % 60

print("Hours:", hours)
print("Minutes:", minutes)
print("Seconds:", remaining_seconds)
```

Part twelvee Explanation:

* Bitwise operators work on binary numbers (0 and 1).

Example:

A = 1
B = 1

A & B → 1 (both must be 1)
A | B → 1 (any one is 1)
A ^ B → 0 (same values give 0)


```python
num = int(input("Enter number: "))

print("Left Shift by 1:", num << 1)
print("Right Shift by 1:", num >> 1)
```
