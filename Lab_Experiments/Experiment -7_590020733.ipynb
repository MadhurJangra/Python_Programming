```python
def find_max_min(numbers):
    max_num = numbers[0]
    min_num = numbers[0]

    for num in numbers:
        if num > max_num:
            max_num = num
        if num < min_num:
            min_num = num

    return max_num, min_num
```

Part One explanation:

* numbers[0]: Assume first element is both max and min initially
* Loop through list:
       
       *  If current number > max → update max
       
       * If current number < min → update min

* Return both values as a tuple

```python
def sum_of_cubes(n):
    total = 0

    for i in range(1, n):
        total += i ** 3

    return total
```

Part two explanation:
* range(1, n): numbers smaller than n
* i ** 3: cube of number
* total +=: keeps adding cubes
* Returns final sum

```python
def print_numbers(n):
    if n == 0:
        return
    print_numbers(n - 1)
    print(n)
```

Part three explanation:
* Base case: stop when n == 0
* Recursive call goes down to 1
* Then prints while coming back (1 → n)

```python
def fibonacci(n):
    if n <= 1:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

def print_fibonacci(n):
    for i in range(n):
        print(fibonacci(i), end=" ")
```

Part four explanation:
* Fibonacci formula:
* f(0)=0, f(1)=1
* f(n)=f(n-1)+f(n-2)
* print_fibonacci(n) prints first n terms

```python
find_max_min_lambda = lambda lst: (
    max(lst),
    min(lst)
)

# Example
print(find_max_min_lambda([10, 6, 8, 90, 12, 56]))
```

Part six explanation:
* max(lst) → largest value
* min(lst) → smallest value
* Returns tuple (max, min)

```python
def add_numbers(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(add_numbers(1, 2, 3, 4))
```

Part sever c part explanation:

* *numbers allows multiple arguments
* Stored as a tuple
