```python
class Student:
    def __init__(self, name, sap_id, phy, chem, maths):
        self.name = name
        self.sap_id = sap_id
        self.marks = [phy, chem, maths]

    def display(self):
        print("Name:", self.name)
        print("SAP ID:", self.sap_id)
        print("Marks (Phy, Chem, Maths):", self.marks)
        print("---------------------------")


students = []

for i in range(3):
    name = input("Enter name: ")
    sap_id = input("Enter SAP ID: ")
    phy = int(input("Physics marks: "))
    chem = int(input("Chemistry marks: "))
    maths = int(input("Maths marks: "))

    s = Student(name, sap_id, phy, chem, maths)
    students.append(s)

for s in students:
    s.display()
```

Part one explanation:
* Class: blueprint to create objects
* __init__() → constructor (called automatically)
* self → refers to current object
* marks stored as list
* Objects stored in list and displayed

```python
class Student:
    def __init__(self, name, sap_id, phy, chem, maths):
        self.name = name
        self.sap_id = sap_id
        self.marks = [phy, chem, maths]

    def display(self):
        print(self.name, self.sap_id, self.marks)

    def marks_percentage(self):
        total = sum(self.marks)
        return total / 3

    def result(self):
        for m in self.marks:
            if m < 40:
                return "Fail"
        return "Pass"


def class_average(students):
    total = 0
    for s in students:
        total += s.marks_percentage()
    return total / len(students)
```

Part two explanation:
* marks_percentage()
* Adds all marks → divides by 3

    🔹 result()

* Checks each subject
* If any mark < 40 → Fail
* Else → Pass
    
    🔹 class_average()

* Takes list of students
* Finds average of all students

```python
class Parent:
    def show(self):
        print("This is Parent")

class Child(Parent):
    def display(self):
        print("This is Child")

c = Child()
c.show()
c.display()
```

Part three part one explanation:
* Child inherits Parent
* Can use Parent methods

```python
class A:
    def funcA(self):
        print("Class A")

class B(A):
    def funcB(self):
        print("Class B")

class C(B):
    def funcC(self):
        print("Class C")

obj = C()
obj.funcA()
obj.funcB()
obj.funcC()
```

Part three part two explanation:
* A → B → C chain
* C can access both A & B

```python
class A:
    def funcA(self):
        print("Class A")

class B:
    def funcB(self):
        print("Class B")

class C(A, B):
    def funcC(self):
        print("Class C")

obj = C()
obj.funcA()
obj.funcB()
```

Part three part Three explanation:
* One child inherits from multiple parents

```python
class Parent:
    def show(self):
        print("Parent method")

class Child(Parent):
    def show(self):
        print("Child method")

obj = Child()
obj.show()
```

Part four explanation:
* Same method name in parent & child
* Child method overrides parent method
* Output → "Child method"

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Point(self.x + other.x, self.y + other.y)

    def display(self):
        print(f"({self.x}, {self.y})")


p1 = Point(10, 20)
p2 = Point(12, 15)

p3 = p1 + p2
p3.display()
```

Part five explanation:
* __add__() → special method for + operator
* Adds x and y separately
* Returns new object
